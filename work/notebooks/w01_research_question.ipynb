## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
import pandas as pd
import numpy as np

# Ensure the data is loaded
csv_path = './flyrank-wk1-run-notebooks/data/raw/content_refresh_anonymized.csv'
try:
    df = pd.read_csv(csv_path)
except Exception:
    import os
    if 'df' in globals():
        df = globals()['df']
    else:
        raise FileNotFoundError("Could not find the dataset.")

# 1. Total 90-day clicks
total_clicks = df['clicks_90d'].sum()

# 2. Slice of content currently declining
declining_mask = df['trend_direction'] == 'down'
num_declining = declining_mask.sum()
pct_declining = (num_declining / len(df)) * 100
clicks_at_risk = df.loc[declining_mask, 'clicks_90d'].sum()
pct_clicks_at_risk = (clicks_at_risk / total_clicks) * 100

print(f"--- Impact Analysis for Decision / Action ---")
print(f"Total Content Items: {len(df):,}")
print(f"Total 90-Day Clicks across portfolio: {total_clicks:,.0f}")
print(f"Number of Declining Content Items (Need Action): {num_declining:,} ({pct_declining:.1f}% of portfolio)")
print(f"90-Day Clicks currently at risk: {clicks_at_risk:,.0f} ({pct_clicks_at_risk:.1f}% of total traffic)")
```

* **What decision does this work improve?**
  This work directly improves the **Content Refresh and Optimization Prioritization** decision. Instead of choosing pages to refresh based on intuition, recency, or arbitrary schedules, we identify which specific pages are experiencing true decay but still possess high market demand.

* **Who acts on it?**
  The primary actors are **SEO Managers, Editorial Directors, and Content Writers** who have limited weekly capacity and must allocate their budgets and writing hours where they will yield the highest recovery return.

* **What does a wrong recommendation cost?**
  A wrong recommendation carries two massive costs:
  1. **Direct Editorial Waste (False Positives):** Spending hours rewriting a page that is stable, seasonal, or facing a search intent shift, resulting in zero recovery of traffic.
  2. **Unmitigated Traffic Decay (False Negatives):** Ignoring a high-value page that is actively decaying.
  
  Our portfolio metrics demonstrate the scale of this risk:
  * Across the **30,000** content items, total organic traffic stands at **482,920 clicks** over 90 days.
  * Currently, **16,262 items (54.2% of the portfolio)** are experiencing a downward trend.
  * This decline puts **217,248 clicks (45.0% of all organic traffic)** directly at risk of complete loss if no action is taken.

Optimizing this selection process with data-driven ranking ensures we protect this 45% traffic baseline while minimizing wasted editorial spend.
